```python
numeric_cols = df.select_dtypes(include="number").columns

results = []

for col in numeric_cols:
    mask = (df[col] > 0) & (df[col] < 0.000001)
    values = df.loc[mask, col]

    if len(values) > 0:
        results.append({
            "Column": col,
            "Small_Value_Count": len(values),
            "Percentage": round(len(values) / len(df) * 100, 2),
            "Minimum_Value": values.min(),
            "Maximum_Value": values.max(),
            "Sample_Values": values.head(5).tolist()
        })


if results:
    result_df = pd.DataFrame(results)
    print(result_df.to_string(index=False))
else:
    print("No values found in the specified range.")
```

```python

# 1. Check whether values look uniformly distributed
cols = [
    "traffic_congestion_level",
    "warehouse_inventory_level",
    "handling_equipment_availability",
    "weather_condition_severity",
    "supplier_reliability_score",
    "cargo_condition_status",
    "driver_behavior_score",
    "fatigue_monitoring_score"
]

print("DISTRIBUTION CHECK")

for col in cols:
    print(f"\n{col}")
    print(df[col].describe(
        percentiles=[0.01, 0.10, 0.25, 0.50, 0.75, 0.90, 0.99]
    ).round(4))

# 2. Correlation with delivery outcomes
target_cols = [
    "eta_variation_hours",
    "delivery_time_deviation",
    "shipping_costs",
    "delay_probability",
    "disruption_likelihood_score"
]

available = [c for c in cols + target_cols if c in df.columns]

print("\nCORRELATION MATRIX")
print(df[available].corr(numeric_only=True).round(3))
```

```python
df["traffic_group"] = pd.cut(
    df["traffic_congestion_level"],
    bins=[-0.01, 3.33, 6.67, 10],
    labels=["Low", "Medium", "High"]
)

traffic_analysis = df.groupby(
    "traffic_group", observed=False
).agg(
    Records=("eta_variation_hours", "count"),
    Avg_ETA_Variation=("eta_variation_hours", "mean"),
    Avg_Delivery_Deviation=("delivery_time_deviation", "mean"),
    Avg_Delay_Probability=("delay_probability", "mean"),
    Avg_Shipping_Cost=("shipping_costs", "mean")
).round(4)

print(traffic_analysis)
```

```python

df["supplier_group"] = pd.cut(
    df["supplier_reliability_score"],
    bins=[-0.001, 0.33, 0.66, 1.001],
    labels=["Low Reliability", "Medium Reliability", "High Reliability"]
)

supplier_analysis = df.groupby(
    "supplier_group", observed=False
).agg(
    Records=("lead_time_days", "count"),
    Avg_Lead_Time=("lead_time_days", "mean"),
    Avg_ETA_Variation=("eta_variation_hours", "mean"),
    Avg_Delay_Probability=("delay_probability", "mean"),
    Avg_Shipping_Cost=("shipping_costs", "mean")
).round(4)

print(supplier_analysis)
```

```python

df["inventory_group"] = pd.cut(
    df["warehouse_inventory_level"],
    bins=[-0.001, 100, 300, 600, 1000.001],
    labels=[
        "Very Low",
        "Low",
        "Medium",
        "High"
    ]
)

inventory_analysis = df.groupby(
    "inventory_group", observed=False
).agg(
    Records=("order_fulfillment_status", "count"),
    Avg_Demand=("historical_demand", "mean"),
    Avg_Loading_Time=("loading_unloading_time", "mean"),
    Avg_Delay_Probability=("delay_probability", "mean"),
    Avg_Lead_Time=("lead_time_days", "mean")
).round(4)

print(inventory_analysis)

print("\nFulfillment status by inventory group:")
print(pd.crosstab(
    df["inventory_group"],
    df["order_fulfillment_status"],
    normalize="index"
).round(3))
```

```python

target = "order_fulfillment_status"

related_cols = [
    "warehouse_inventory_level",
    "historical_demand",
    "loading_unloading_time",
    "handling_equipment_availability",
    "lead_time_days",
    "delay_probability",
    "disruption_likelihood_score"
]


print("Correlation with order_fulfillment_status:")

print(
    df[[target] + related_cols]
    .corr(numeric_only=True)[target]
    .drop(target)
    .sort_values()
    .round(4)
)


df["fulfillment_score_group"] = pd.cut(
    df[target],
    bins=[-0.001, 0.33, 0.66, 1.001],
    labels=["Low Score", "Medium Score", "High Score"]
)

print("\nGroup comparison:")

print(
    df.groupby("fulfillment_score_group", observed=False)
    .agg(
        Records=(target, "count"),
        Avg_Inventory=("warehouse_inventory_level", "mean"),
        Avg_Demand=("historical_demand", "mean"),
        Avg_Lead_Time=("lead_time_days", "mean"),
        Avg_Loading_Time=("loading_unloading_time", "mean"),
        Avg_Delay_Probability=("delay_probability", "mean")
    )
    .round(4)
)
```

```python



business_cols = [
    "traffic_congestion_level",
    "fuel_consumption_rate",
    "eta_variation_hours",
    "shipping_costs",
    "supplier_reliability_score",
    "lead_time_days",
    "weather_condition_severity",
    "route_risk_level",
    "disruption_likelihood_score",
    "delay_probability",
    "delivery_time_deviation"
]

corr = df[business_cols].corr(numeric_only=True)


pairs = corr.where(
    np.triu(np.ones(corr.shape), k=1).astype(bool)
).stack().reset_index()

pairs.columns = ["Variable_1", "Variable_2", "Correlation"]

pairs["Absolute_Correlation"] = pairs["Correlation"].abs()

print(
    pairs.sort_values("Absolute_Correlation", ascending=False)
    .head(15)
    .round(4)
    .to_string(index=False)
)
```
